```python
import pandas as pd

df1 = pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2009-2010')
df2 = pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2010-2011')
df = pd.concat([df1, df2], ignore_index=True)

print(df.shape)
df.info()
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
import os
print(os.getcwd())
```

```python
df['Invoice'] = df['Invoice'].astype(str)
cancelled = df[df['Invoice'].str.startswith('C')]
print(cancelled.shape)
cancelled[['Invoice', 'Quantity', 'Price']].describe()
```

```python
negative_price = df[df['Price'] < 0]
print(negative_price.shape)
negative_price[['Invoice', 'StockCode', 'Description', 'Quantity', 'Price']].head(10)
```

```python
df_clean = df[
    (~df['Invoice'].str.startswith('C')) &
    (~df['Invoice'].str.startswith('A')) &
    (df['Customer ID'].notna()) &
    (df['Quantity'] > 0) &
    (df['Price'] > 0)
].copy()

print("Original rows:", df.shape[0])
print("Cleaned rows:", df_clean.shape[0])
print("Rows removed:", df.shape[0] - df_clean.shape[0])

df_clean.isnull().sum()
```

```python
df_clean.to_csv('../data/processed/online_retail_cleaned.csv', index=False)
print("Saved.")
```
